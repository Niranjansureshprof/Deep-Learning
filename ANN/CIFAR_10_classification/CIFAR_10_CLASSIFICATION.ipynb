## Step 1: Import Libraries

```python
import numpy as np
import keras
from keras.models import Sequential
from keras.layers import Input,Dense,Flatten
from PIL import Image
from keras.models import load_model
```

## Step 2: Import Dataset

```python
from keras.datasets import cifar10
```

## Step 3: Data Understanding

```python
(x_train,y_train),(x_test,y_test)=keras.datasets.cifar10.load_data()
```

```python
x_train.shape,y_train.shape
```

## Step 4: Data Preparation

```python
x_train=x_train/255.0
```

```python
x_test=x_test/255.0
```

```python
x_train.min(),x_train.max()
```

```python
x_test.min(),x_test.max()
```

## Step 5: Model Building 

```python
model=Sequential()
model.add(Input(shape=(32,32,3)))
model.add(Flatten())
model.add(Dense(units=100,activation="relu"))
model.add(Dense(units=10,activation="softmax"))
```

## Step 6: Model compilation

```python
model.compile(optimizer='rmsprop',loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

## Step 7: Model Training || Step 8: Model Testing || Step 9: Model Evaluation

```python
model.fit(x_train,y_train,batch_size=50,epochs=10,validation_data=(x_test,y_test))
```

```python
model.summary()
```

## Let's build model intelligence

## Step 1: get the model intelligence in h5 format

```python
model.save("classifier.h5")
```

## Step 2: Image preprocessing

```python
image_path="airplane-png-27947.png"
image=Image.open(image_path).convert("RGB").resize(size=(32,32))
image=np.array(image)
image.shape
```

## Step 3: prediction

```python
prediction_model=load_model("classifier.h5")
```

```python
def predict_image(image):
    image=image/255.0
    image=np.expand_dims(a=image,axis=0)
    prediction=prediction_model.predict(image)
    return np.argmax(prediction)
    
```

```python
predict_image(image)
```
